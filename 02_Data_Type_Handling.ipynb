## 1. Identifying Data Types

A data type tells Python how a value is represented and what operations are suitable.

Use `df.dtypes` to inspect column types and `df.info()` for a summary.

- **Simple example:** `25` is an integer; `25.5` is a float; `"25"` is a string.
- **Real-world example:** A hospital dataset may contain patient age, diagnosis category, admission date, and consent status.
- **Business example:** Sales data may contain quantity, revenue, product category, and order date.
- **AI/ML use case:** A preprocessing pipeline treats numeric and categorical features differently.
- **Why required / problem solved:** Incorrect types can cause calculation errors, failed date operations, incorrect sorting, or unsuitable model inputs.

```python
df = pd.DataFrame({
    "age": [25, 30, 42],
    "salary": [25000.5, 42000.0, 51000.75],
    "department": ["Sales", "IT", "HR"],
    "is_active": [True, False, True],
    "joining_date": ["2024-01-10", "2023-06-15", "2022-09-01"]
})
print(df.dtypes)
df.info()
```

## 2. Numerical Data

**Definition:** Numerical data represents measurable or countable quantities. It can be integer or floating-point data.

- **Simple example:** Number of books = `4`; height = `165.5`.
- **Real-world example:** Patient temperature or distance travelled.
- **Business example:** Product quantity, sales revenue, and delivery cost.
- **AI/ML use case:** Age and income may be numerical model features.
- **Why required:** Numerical values support arithmetic, statistics, scaling, and many model algorithms.

**Caution:** An ID such as `10025` may look numeric but usually represents an identifier, not a quantity to average.

```python
numbers = pd.DataFrame({
    "quantity": [2, 5, 10],
    "price": [99.5, 120.0, 75.25]
})
numbers["total"] = numbers["quantity"] * numbers["price"]
display(numbers)
print(numbers.dtypes)
```

## 3. Categorical Data

**Definition:** Categorical data represents groups or labels, such as department, city, or product type.

- **Simple example:** `"Red"`, `"Blue"`, `"Green"`.
- **Real-world example:** Blood group categories or vehicle type.
- **Business example:** Payment method such as Cash, Card, or UPI.
- **AI/ML use case:** One-hot encoding can convert categories into numeric model inputs.
- **Why required:** Treating labels as ordinary numbers can incorrectly imply a mathematical order or distance.

Some categories are nominal (no natural order); others are ordinal (a meaningful order exists).

```python
products = pd.DataFrame({
    "product": ["Phone", "Laptop", "Phone", "Tablet"],
    "size": ["Small", "Large", "Medium", "Large"]
})
products["product"] = products["product"].astype("category")
print(products.dtypes)
print(products["product"].cat.categories)
```

## 4. Boolean Data

**Definition:** Boolean data has two logical values: `True` and `False`.

- **Simple example:** `is_pass = True`.
- **Real-world example:** Whether a patient has completed a form.
- **Business example:** Whether an order has been delivered.
- **AI/ML use case:** A binary flag can represent whether a customer used a promotion.
- **Why required:** Boolean types make filtering and logical conditions clear and reduce ambiguity.

```python
orders = pd.DataFrame({
    "order_id": [1, 2, 3],
    "delivered": [True, False, True]
})
print(orders[orders["delivered"]])
print(orders["delivered"].dtype)
```

## 5. Date/Time Data

**Definition:** Date/time data represents calendar dates, times, or timestamps.

- **Simple example:** `2026-11-10`.
- **Real-world example:** A patient's admission and discharge timestamps.
- **Business example:** Order date, payment time, and delivery date.
- **AI/ML use case:** Extract day, month, weekday, or elapsed time for forecasting.
- **Why required:** Correct datetime types allow date arithmetic, chronological sorting, time filtering, and feature extraction.

```python
dates = pd.DataFrame({"order_date": ["2026-01-05", "2026-02-18", "2026-03-22"]})
dates["order_date"] = pd.to_datetime(dates["order_date"], errors="coerce")
dates["month"] = dates["order_date"].dt.month
dates["weekday"] = dates["order_date"].dt.day_name()
display(dates)
print(dates.dtypes)
```

## 6. String Data

**Definition:** String data is text: names, descriptions, codes, and labels.

- **Simple example:** `"Bilal"`.
- **Real-world example:** A patient's address or doctor's name.
- **Business example:** Product description or customer feedback.
- **AI/ML use case:** Text may be cleaned and transformed into features using NLP techniques.
- **Why required:** Text operations such as trimming spaces, changing case, searching, and extracting patterns require appropriate string handling.

A numeric-looking string can be intentional, for example a postcode or account code. Do not convert it to a number if leading zeros or its identifier meaning matter.

```python
names = pd.Series(["  bilal  ", "ASHA", "ravi kumar"])
clean_names = names.str.strip().str.title()
print(clean_names.tolist())
```

## 7. Converting Data Types

Use conversion only when it matches the meaning of the data.

| Conversion | Example | Typical tool |
|---|---|---|
| String → Numeric | `"250"` → `250` | `pd.to_numeric()` |
| String → Date | `"2026-11-10"` → timestamp | `pd.to_datetime()` |
| Numeric → Categorical | `1`, `2`, `3` as group codes | `astype("category")` |
| Integer → Float | `25` → `25.0` | `astype(float)` |

- **Simple example:** Convert a price column stored as text before calculating totals.
- **Real-world example:** Convert timestamp text from a machine log into datetime values.
- **Business example:** Convert a sales amount to numeric before calculating monthly revenue.
- **AI/ML use case:** Convert feature columns into types accepted by the preprocessing steps.
- **Why required:** It prevents type-related errors and ensures operations reflect the meaning of the values.

## 8. `astype()`

**Definition:** `astype()` converts a pandas Series or DataFrame column to a specified dtype.

Use it when the values are already compatible with the requested type. Invalid values may cause an error.

```python
df = pd.DataFrame({
    "age_text": ["21", "30", "45"],
    "rating": [4, 5, 3]
})
df["age"] = df["age_text"].astype(int)
df["rating_float"] = df["rating"].astype(float)
df["rating_category"] = df["rating"].astype("category")
display(df)
print(df.dtypes)
```

## 9. `to_numeric()`

**Definition:** `pd.to_numeric()` converts values to numeric types.

`errors="coerce"` changes values that cannot be converted into `NaN`, allowing you to inspect them safely.

- **Simple example:** `"500"` becomes `500`; `"unknown"` becomes missing.
- **Real-world example:** Convert numeric readings received as text from a device.
- **Business example:** Convert sales amounts imported from a CSV file.
- **AI/ML use case:** Ensure a numeric feature is numeric before imputing or scaling it.
- **Why required:** It handles mixed or messy numeric columns more safely than direct conversion.

```python
sales = pd.DataFrame({"amount": ["1000", "2500.50", "unknown", " 300 "]})
sales["amount_numeric"] = pd.to_numeric(sales["amount"].str.strip(), errors="coerce")
display(sales)
print("Invalid or missing conversions:")
display(sales[sales["amount_numeric"].isna()])
```

## 10. `to_datetime()`

**Definition:** `pd.to_datetime()` converts date/time values into pandas datetime values.

- **Simple example:** Convert `"10-11-2026"` to a date.
- **Real-world example:** Parse timestamps from equipment logs.
- **Business example:** Convert order-date text to datetime before monthly sales analysis.
- **AI/ML use case:** Extract time-based features for demand or traffic forecasting.
- **Why required:** Text dates cannot reliably support date arithmetic or datetime-specific features.

When dates are ambiguous, specify the expected format explicitly.

```python
date_data = pd.DataFrame({
    "date_text": ["10-11-2026", "15-11-2026", "not-a-date"]
})
date_data["date"] = pd.to_datetime(
    date_data["date_text"],
    format="%d-%m-%Y",
    errors="coerce"
)
display(date_data)
```

## 11. Handling Invalid Conversions

**Definition:** Invalid conversions happen when a value cannot be represented in the requested type.

Common strategies:
- Use `errors="coerce"` to mark invalid values as missing.
- Inspect the affected rows.
- Correct values only when the intended value can be verified.
- Keep an audit of how many values were affected.
- Decide whether to impute, exclude, or retain missing values based on the task.

- **Simple example:** `"abc"` cannot be converted to a number.
- **Real-world example:** A device sends `"error"` instead of a temperature reading.
- **Business example:** A spreadsheet contains `"N/A"` in a price column.
- **AI/ML use case:** Invalid numeric values are identified before a model's imputation step.
- **Why required:** Silently ignoring invalid data can hide data quality problems.

```python
raw_values = pd.DataFrame({"quantity": ["5", "10", "three", "-2", None]})
raw_values["quantity_num"] = pd.to_numeric(raw_values["quantity"], errors="coerce")

invalid_conversion = raw_values["quantity"].notna() & raw_values["quantity_num"].isna()
business_rule_invalid = raw_values["quantity_num"] < 0

print("Rows that failed numeric conversion:")
display(raw_values[invalid_conversion])
print("Rows violating non-negative quantity rule:")
display(raw_values[business_rule_invalid])
print("Failed conversion count:", int(invalid_conversion.sum()))
```

## 12. Detecting Incorrect Data Types

Use `dtypes`, `info()`, and knowledge of the dataset to identify suspicious columns.

Questions to ask:
- Is a salary stored as `object` or string?
- Is a date stored as plain text?
- Is a customer ID being treated as a measurable quantity?
- Are `True`/`False` values stored as inconsistent strings?
- Does the column's type match its business meaning?

- **Simple example:** A price column has dtype `object` because it contains `"₹500"`.
- **Real-world example:** An event timestamp is imported as text.
- **Business example:** A sales report cannot calculate revenue because amounts include currency symbols and commas.
- **AI/ML use case:** Incorrect types can cause preprocessing or model fitting to fail.
- **Why required:** Type inspection catches problems before calculations and modelling.

```python
mixed = pd.DataFrame({
    "price": ["₹1,200", "₹850", "₹2,000"],
    "signup_date": ["2026/01/01", "2026/02/15", "2026/03/20"],
    "active": ["Yes", "No", "Yes"],
    "customer_id": ["001", "002", "003"]
})
print("Before conversion:")
print(mixed.dtypes)

# Remove currency symbol and thousands separator, then convert.
mixed["price"] = pd.to_numeric(
    mixed["price"].str.replace("₹", "", regex=False).str.replace(",", "", regex=False),
    errors="coerce"
)
mixed["signup_date"] = pd.to_datetime(mixed["signup_date"], errors="coerce")
mixed["active"] = mixed["active"].map({"Yes": True, "No": False}).astype("boolean")

# Keep customer_id as text because leading zeros are meaningful.
print("\nAfter conversion:")
print(mixed.dtypes)
display(mixed)
```

## 13. Four Required Conversion Examples

### A. String → Numeric
`"100"` becomes `100`. Use `to_numeric()` and inspect failed conversions.

### B. String → Date
`"2026-11-10"` becomes a datetime value. Use `to_datetime()` with a known format where possible.

### C. Numeric → Categorical
A numeric code such as `1`, `2`, `3` may represent groups, not quantities. Convert it to category when that matches its meaning.

### D. Integer → Float
`25` becomes `25.0`. Use `astype(float)` when a decimal-capable representation is needed.

```python
conversions = pd.DataFrame({
    "number_text": ["100", "250", "400"],
    "date_text": ["2026-01-01", "2026-01-02", "2026-01-03"],
    "group_code": [1, 2, 1],
    "whole_number": [10, 20, 30]
})

conversions["number"] = pd.to_numeric(conversions["number_text"])
conversions["date"] = pd.to_datetime(conversions["date_text"])
conversions["group_code"] = conversions["group_code"].astype("category")
conversions["decimal_number"] = conversions["whole_number"].astype(float)

display(conversions)
print(conversions.dtypes)
```

## 14. Why Correct Data Types Are Important

Correct data types help ensure:

1. Correct calculations – numeric columns can be added and averaged.
2. Correct date operations – dates can be sorted and compared chronologically.
3. Correct text handling – names and descriptions can be cleaned safely.
4. Correct category handling – labels are not treated as continuous measurements.
5. Better memory use – suitable pandas dtypes can reduce memory for some datasets.
6. Reliable machine learning – preprocessing steps receive the expected input types.

**Business example:** If sales revenue is stored as text, total revenue calculations may fail or produce incorrect results.

**AI/ML example:** A category such as `"Gold"` should usually be encoded as a category, not interpreted as a numeric measurement.

**Important caution:** Choose types based on meaning, not appearance. A postcode like `"00501"` should normally remain text so its leading zeros are preserved.

## Final Summary

- Identify the meaning and current dtype of each column.
- Distinguish numerical, categorical, Boolean, datetime, and string data.
- Use `astype()` for compatible conversions.
- Use `pd.to_numeric()` for numeric conversion and `pd.to_datetime()` for date conversion.
- Use `errors="coerce"` carefully, then inspect the invalid values.
- Preserve identifiers such as postcodes and customer IDs as strings when formatting matters.
- Correct data types make analysis and machine learning more reliable.
